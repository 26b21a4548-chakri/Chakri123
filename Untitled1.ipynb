```python
# ============================================================
# BRIDGE LOAD DISTRIBUTION CALCULATOR - B.TECH MINI PROJECT
# ============================================================
#
# Mathematical concept:
#       AX = B
#       X = A^(-1)B
#
# Real-world context:
#       Simplified educational model of load distribution
#       among three bridge supports.
#
# IMPORTANT ASSUMPTION:
# A real 3-support continuous bridge is generally statically
# indeterminate. Therefore, this mini-project uses an explicit
# simplified linear load-distribution model. The matrix A
# represents the assumed support-load influence relationship.
#
# This is for mathematical/educational demonstration and is
# NOT a structural-engineering design calculation.
# ============================================================


# -----------------------------
# 1. IMPORT LIBRARIES
# -----------------------------

import numpy as np
import pandas as pd
import sympy as sp
import matplotlib.pyplot as plt

print("=" * 70)
print("        BRIDGE LOAD DISTRIBUTION CALCULATOR")
print("=" * 70)


# -----------------------------
# 2. USER INPUTS
# -----------------------------

# Three bridge supports (metres)
support_positions = np.array([
    0.0,
    10.0,
    20.0
])

# Three loads (kN)
loads = np.array([
    100.0,
    150.0,
    80.0
])

# Initial positions of the loads (metres)
load_positions_before = np.array([
    4.0,
    10.0,
    16.0
])

# Truck = second load
truck_index = 1

# Truck moves from 10 m to 14 m
truck_new_position = 14.0


# -----------------------------
# 3. MATHEMATICAL MODEL
# -----------------------------
#
# We use a simple, symmetric 3-support influence matrix.
#
#              [ 2  -1   0 ]
#      A =     [ -1  2  -1 ]
#              [ 0  -1   2 ]
#
# X = [R1 R2 R3]^T
#
# B represents the effect of the external loads.
#
# The load influence is calculated from the position of each
# load relative to the bridge supports.
# -----------------------------

A = np.array([
    [2.0, -1.0, 0.0],
    [-1.0, 2.0, -1.0],
    [0.0, -1.0, 2.0]
])


def calculate_load_vector(load_positions, loads, supports):
    """
    Create B for AX = B.

    This is a simplified educational load-distribution model.
    """

    x1, x2, x3 = supports

    # Influence of each load on the support regions.
    B1 = np.sum(
        loads * (x3 - load_positions) / (x3 - x1)
    )

    B3 = np.sum(
        loads * (load_positions - x1) / (x3 - x1)
    )

    total_load = np.sum(loads)

    # Middle-support influence is the remaining load effect.
    B2 = total_load - B1 - B3

    B = np.array([
        B1,
        B2,
        B3
    ])

    return B


# -----------------------------
# 4. BEFORE TRUCK MOVEMENT
# -----------------------------

B_before = calculate_load_vector(
    load_positions_before,
    loads,
    support_positions
)


# -----------------------------
# 5. CHECK UNIQUE SOLUTION
# -----------------------------

det_A = np.linalg.det(A)

print("\n1. MATRIX A")
print("-" * 70)
print(A)

print("\nDeterminant of A =", round(det_A, 6))

if abs(det_A) < 1e-10:
    raise ValueError(
        "Matrix A is singular. A unique solution cannot be obtained."
    )

print("✓ det(A) is non-zero.")
print("✓ A is invertible.")
print("✓ Unique solution exists.")


# -----------------------------
# 6. MATRIX INVERSE
# -----------------------------

A_sym = sp.Matrix(A)
A_inverse_sym = A_sym.inv()

print("\n2. MATRIX INVERSE A^(-1)")
print("-" * 70)
sp.pprint(A_inverse_sym)


# -----------------------------
# 7. SOLVE AX = B
# -----------------------------

# Method 1: NumPy
reactions_before = np.linalg.solve(
    A,
    B_before
)

# Method 2: Matrix inverse
A_inverse = np.linalg.inv(A)

reactions_inverse = A_inverse @ B_before

print("\n3. VECTOR B — BEFORE TRUCK MOVEMENT")
print("-" * 70)
print(B_before)

print("\n4. REACTIONS — BEFORE TRUCK MOVEMENT")
print("-" * 70)

for i, reaction in enumerate(reactions_before):
    print(
        f"Support {i+1}: {reaction:.2f} kN"
    )

print("\nUsing A^(-1)B:")
print(reactions_inverse)

difference = np.max(
    np.abs(reactions_before - reactions_inverse)
)

print(
    f"\nDifference between both methods: "
    f"{difference:.10f}"
)


# -----------------------------
# 8. MOVE THE TRUCK
```

```python
# ============================================================
# BRIDGE LOAD DISTRIBUTION CALCULATOR
# B.Tech Mathematics Mini Project
# ============================================================

import numpy as np
import sympy as sp
import matplotlib.pyplot as plt

# ------------------------------------------------------------
# 1. PROJECT TITLE
# ------------------------------------------------------------

print("=" * 75)
print("          BRIDGE LOAD DISTRIBUTION CALCULATOR")
print("          B.Tech Mathematics Mini Project")
print("=" * 75)


# ------------------------------------------------------------
# 2. BRIDGE DATA
# ------------------------------------------------------------

# Support positions in metres
supports = np.array([0.0, 10.0, 20.0])

# Loads in kN
loads = np.array([100.0, 150.0, 80.0])

# Initial load positions in metres
load_positions_before = np.array([4.0, 10.0, 16.0])

# Truck = second load
truck_index = 1

# Truck moves from 10 m to 14 m
truck_new_position = 14.0


# ------------------------------------------------------------
# 3. MATRIX A
# ------------------------------------------------------------

A = np.array([
    [ 2.0, -1.0,  0.0],
    [-1.0,  2.0, -1.0],
    [ 0.0, -1.0,  2.0]
])


# ------------------------------------------------------------
# 4. FUNCTION TO CALCULATE B
# ------------------------------------------------------------

def calculate_B(load_positions, loads, supports):

    x1, x2, x3 = supports

    # Load influence on left support
    B1 = np.sum(
        loads * (x3 - load_positions) / (x3 - x1)
    )

    # Load influence on right support
    B3 = np.sum(
        loads * (load_positions - x1) / (x3 - x1)
    )

    # Remaining load influence
    total_load = np.sum(loads)

    B2 = total_load - B1 - B3

    return np.array([B1, B2, B3])


# ------------------------------------------------------------
# 5. BEFORE TRUCK MOVEMENT
# ------------------------------------------------------------

B_before = calculate_B(
    load_positions_before,
    loads,
    supports
)


# ------------------------------------------------------------
# 6. MATRIX CHECK
# ------------------------------------------------------------

det_A = np.linalg.det(A)

print("\nMATRIX A")
print("-" * 75)
print(A)

print("\nDeterminant of A =", round(det_A, 4))

if abs(det_A) < 1e-10:
    raise ValueError("Matrix A is singular.")

print("✓ Matrix A is invertible.")
print("✓ Unique solution exists.")


# ------------------------------------------------------------
# 7. MATRIX INVERSE
# ------------------------------------------------------------

A_sym = sp.Matrix(A)
A_inverse_sym = A_sym.inv()

print("\nMATRIX A⁻¹")
print("-" * 75)
sp.pprint(A_inverse_sym)


# ------------------------------------------------------------
# 8. SOLVE AX = B
# ------------------------------------------------------------

reactions_before = np.linalg.solve(
    A,
    B_before
)

print("\nLOAD VECTOR B — BEFORE")
print("-" * 75)
print(B_before)

print("\nSUPPORT REACTIONS — BEFORE TRUCK MOVEMENT")
print("-" * 75)

for i, r in enumerate(reactions_before):
    print(f"Support {i+1}: {r:.2f} kN")


# ------------------------------------------------------------
# 9. MOVE TRUCK
# ------------------------------------------------------------

load_positions_after = load_positions_before.copy()

load_positions_after[truck_index] = truck_new_position


# ------------------------------------------------------------
# 10. AFTER
```
